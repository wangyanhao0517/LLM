# 二叉搜索树（BST）


1.   求路径、求范围？ **->** 用pre-order，“自顶向下” + 区间传递
2.   求第k大 / 小、找顺序错误、恢复数据 **->** 使用In-order
3.   求子树信息、求高度 **->** post-order 后序遍历 + “自底向上”
